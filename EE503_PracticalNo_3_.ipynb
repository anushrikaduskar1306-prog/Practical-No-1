```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
file = "/content/communication_system_fault__dataset.csv"
data_raw = pd.read_csv(file, header=None)

# Remove outer quotes and split each string by comma
processed_data = data_raw.iloc[:, 0].str.strip('"').str.split(',', expand=True)

# Set the first row as header
data = processed_data.rename(columns=processed_data.iloc[0]).drop(processed_data.index[0])

print(data.columns)
```

```python
print(data.isnull().sum())
```
